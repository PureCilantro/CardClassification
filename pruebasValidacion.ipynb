```python
import os
import cv2
import numpy as np
import pandas as pd
from keras.models import load_model
import matplotlib.pyplot as plt
```

```python
img_path = 'input/valid'
Sx = 224
Sy = 224

df = pd.read_csv('input/cards.csv')
category_labels = df.labels.value_counts().index
categories = {}
for index,cat in enumerate(category_labels) :
  categories[cat]=index
  
id_to_category = {v: k for k, v in categories.items()}
```

```python
modelo_final = load_model('modelo_final.h5')
```

```python
for file in os.listdir(img_path):
    full_path = os.path.join(img_path, file)
    
    img = cv2.imread(full_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (Sx, Sy))
    img_norm = img.astype('float32') / 255.0
    img_input = img_norm.reshape(-1,Sx,Sy,3)
    
    pred = modelo_final.predict(img_input)
    pred_class = np.argmax(pred, axis=1)[0]
    pred_label = id_to_category[pred_class]
    
    plt.figure(figsize=(3, 3))
    plt.imshow(img)
    plt.title(f'Prediction: {pred_label}')
    plt.axis('off')
    plt.show()
```
